```python
salary = df[[
    "Територіальний розріз",
    "Територіальний розріз.1",
    "Період",
    "Значення cпостереження"
]].copy()

salary.head()
```

```python
salary = salary.rename(columns={
    "Територіальний розріз": "region_code",
    "Територіальний розріз.1": "region",
    "Період": "year",
    "Значення cпостереження": "salary_uah"
})

salary.head()
```

```python
salary.isna().sum()
```

```python
salary["salary_uah"].describe()
```

```python
salary[
    salary["region"].isin([
        "Автономна Республіка Крим",
        "Севастополь"
    ])
][["region", "year", "salary_uah"]]
```

```python
salary = salary.dropna(subset=["salary_uah"]).copy()
```

```python
salary.isna().sum()
```

```python
salary_ukraine = salary[salary["region"] == "Україна"].copy()

salary_regions = salary[salary["region"] != "Україна"].copy()
```

```python
print("Украина:")
print(salary_ukraine[["year", "salary_uah"]])

print("\nКоличество региональных наблюдений:")
print(salary_regions.groupby("year").size())
```

```python
salary_ukraine[["year", "salary_uah"]]
```

```python
regional = salary_regions[
    salary_regions["year"].isin([2021, 2024])
].copy()

regional.head()
```

```python
regional_pivot = regional.pivot(
    index="region",
    columns="year",
    values="salary_uah"
)

regional_pivot.head()
```

```python
regional_pivot["change_pct"] = (
    (regional_pivot[2024] - regional_pivot[2021])
    / regional_pivot[2021] * 100
)

regional_pivot.sort_values("change_pct", ascending=False)
```

```python
cv_2021 = (
    regional_pivot[2021].std()
    / regional_pivot[2021].mean()
    * 100
)

cv_2024 = (
    regional_pivot[2024].std()
    / regional_pivot[2024].mean()
    * 100
)

print("CV 2021:", cv_2021)
print("CV 2024:", cv_2024)
```

```python
import matplotlib.pyplot as plt
plot_data = regional_pivot.sort_values(2024)

plt.figure(figsize=(10, 8))

plt.barh(
    plot_data.index,
    plot_data[2024]
)

plt.xlabel("Середньомісячна заробітна плата, грн")
plt.ylabel("Регіон")
plt.title("Середньомісячна заробітна плата за регіонами у 2024 році")

plt.show()
```
